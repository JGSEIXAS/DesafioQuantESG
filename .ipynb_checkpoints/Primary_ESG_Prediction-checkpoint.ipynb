# FUNÇÕES DO SISTEMA


```python
if __name__ == '__main__':
    lista_de_ativos_para_teste = ['VALE3.SA', 'ITUB4.SA', 'LREN3.SA']
    
    DATA_FIM_TREINO = datetime(2023, 12, 31).date()
    DATA_FIM_VALIDACAO = datetime(2024, 9, 30).date()
    
    for ativo_alvo in lista_de_ativos_para_teste:
        
        print("\n" + "="*80)
        print(f" INICIANDO ANÁLISE PARA: {ativo_alvo} ".center(80, "="))
        print("="*80)
        
        config = CONFIGURACAO_ATIVOS.get(ativo_alvo, CONFIGURACAO_ATIVOS["DEFAULT"])
        print(f"Setor: {config['setor']}")

        start_date = DATA_FIM_VALIDACAO - timedelta(days=5*356) # Corrigido para 356 para evitar overlap
        end_date = datetime.now()
        fatores_base_tickers = ['CL=F', 'BRL=X']

        dados_brutos = coletar_dados(
            ticker_principal=ativo_alvo,
            fatores_base=fatores_base_tickers,
            fatores_especificos=config['fatores_especificos'],
            nomes_fatores=config['nomes_fatores'],
            start=start_date,
            end=end_date
        )
        
        if dados_brutos:
            df_modelo_pronto = preparar_dataframe(dados_brutos)
            
            if df_modelo_pronto is not None:
                colunas_treino = ['Momentum_5d', 'Volatilidade_10d', 'Score_ESG_Politico', 'Preco_Petroleo'] + config['nomes_fatores']
                
                modelo, datasets = treinar_avaliar_modelo(df_modelo_pronto, colunas_treino, DATA_FIM_TREINO, DATA_FIM_VALIDACAO)
                
                executar_backtest(modelo, df_modelo_pronto, datasets, ticker_principal=ativo_alvo)
```

```python
# ====================================================================
# CÓDIGO UNIFICADO E COMPLETO
# Copie e cole todo este bloco em uma única célula e execute.
# ====================================================================

# --- ETAPA 1: INSTALAÇÃO E IMPORTAÇÕES ---
!pip install yfinance pandas scikit-learn tensorflow matplotlib --user

import yfinance as yf
import pandas as pd
from datetime import datetime, timedelta
import numpy as np
import os
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import tensorflow as tf

# --- ETAPA 2: CONFIGURAÇÃO DOS ATIVOS ---
CONFIGURACAO_ATIVOS = {
    "VALE3.SA": {
        "setor": "Mineração",
        "fatores_especificos": ['SI=F', 'MCHI'], "nomes_fatores": ['Preco_Minerio_Ferro', 'Preco_MCHI']
    },
    "ITUB4.SA": {
        "setor": "Bancário",
        "fatores_especificos": ['^BVSP', 'EWZ'], "nomes_fatores": ['Indice_Ibovespa', 'ETF_Brasil']
    },
    "LREN3.SA": {
        "setor": "Varejo",
        "fatores_especificos": ['^BVSP', 'EWZ'], "nomes_fatores": ['Indice_Ibovespa', 'ETF_Brasil']
    },
    "DEFAULT": {
        "setor": "Geral",
        "fatores_especificos": ['^BVSP', 'EWZ'], "nomes_fatores": ['Indice_Ibovespa', 'ETF_Brasil']
    }
}

# --- ETAPA 3: DEFINIÇÃO DAS FUNÇÕES DO SISTEMA ---
def coletar_dados(ticker_principal, fatores_base, fatores_especificos, nomes_fatores, start, end):
    print(f"--- Coletando dados para {ticker_principal} ---")
    todos_tickers = [ticker_principal] + fatores_base + fatores_especificos
    dados_completos = yf.download(todos_tickers, start=start, end=end, progress=False)
    if dados_completos.empty:
        print(f"Erro: Falha ao baixar dados.")
        return None
    dados_brutos = {'principal': dados_completos['Close'][ticker_principal].to_frame(name='Preco_Fechamento')}
    dados_brutos['petroleo'] = dados_completos['Close']['CL=F'].to_frame(name='Preco_Petroleo')
    dados_brutos['cambio'] = dados_completos['Close']['BRL=X'].to_frame(name='Taxa_Cambio_USD_BRL')
    for i, ticker_fator in enumerate(fatores_especificos):
        nome_coluna = nomes_fatores[i]
        if ticker_fator in dados_completos['Close'].columns:
            dados_brutos[nome_coluna] = dados_completos['Close'][ticker_fator].to_frame(name=nome_coluna)
        else:
            dados_brutos[nome_coluna] = pd.DataFrame(columns=[nome_coluna])
    print("Coleta de dados finalizada.")
    return dados_brutos

def preparar_dataframe(dados_brutos):
    print("--- Realizando engenharia de fatores ---")
    if dados_brutos is None: return None
    df_final = dados_brutos['principal']
    df_final.index = df_final.index.date
    df_final['Momentum_5d'] = df_final['Preco_Fechamento'].pct_change(periods=5)
    df_final['Volatilidade_10d'] = df_final['Preco_Fechamento'].pct_change().rolling(window=10).std()
    cambio_df = dados_brutos['cambio']
    cambio_df.index = cambio_df.index.date
    cambio_df['Score_ESG_Politico'] = cambio_df['Taxa_Cambio_USD_BRL'].pct_change()
    df_final = df_final.merge(cambio_df[['Score_ESG_Politico']], left_index=True, right_index=True, how='left').ffill()
    for nome, df in dados_brutos.items():
        if nome not in ['principal', 'cambio']:
            df.index = df.index.date
            df_final = df_final.merge(df, left_index=True, right_index=True, how='left').ffill()
    df_final['Preco_Amanha'] = df_final['Preco_Fechamento'].shift(-1)
    df_final['Retorno_Futuro'] = (df_final['Preco_Amanha'] - df_final['Preco_Fechamento']) / df_final['Preco_Fechamento']
    df_final['Sinal_Trade'] = 0
    df_final.loc[df_final['Retorno_Futuro'] > 0.001, 'Sinal_Trade'] = 1
    df_modelo = df_final.dropna()
    if df_modelo.empty:
        print("Erro: DataFrame ficou vazio após a limpeza.")
        return None
    print("DataFrame pronto para o treinamento.")
    return df_modelo

def treinar_avaliar_modelo(df_modelo, colunas_fatores, data_fim_treino, data_fim_validacao):
    print("--- Iniciando treinamento do modelo de IA ---")
    if df_modelo is None: return None, None

    tf.config.experimental.enable_op_determinism()
    SEED_VALUE = 42
    os.environ['PYTHONHASHSEED'] = str(SEED_VALUE)
    np.random.seed(SEED_VALUE)
    tf.random.set_seed(SEED_VALUE)

    X = df_modelo[colunas_fatores]
    Y = df_modelo['Sinal_Trade']
    scaler = StandardScaler()
    
    indice_corte_treino = df_modelo.index.searchsorted(data_fim_treino, side='right')
    indice_corte_validacao = df_modelo.index.searchsorted(data_fim_validacao, side='right')

    X_escalado = scaler.fit_transform(X)
    X_treino = X_escalado[:indice_corte_treino]
    Y_treino = Y[:indice_corte_treino]
    X_validacao = X_escalado[indice_corte_treino:indice_corte_validacao]
    Y_validacao = Y[indice_corte_treino:indice_corte_validacao]
    X_teste = X_escalado[indice_corte_validacao:]
    Y_teste = Y[indice_corte_validacao:]
    
    if len(X_treino) == 0 or len(X_validacao) == 0 or len(X_teste) == 0:
        print("Erro: Um dos conjuntos de dados (treino, validação ou teste) está vazio. Ajuste as datas.")
        return None, None
        
    modelo = Sequential([
        Dense(128, activation='relu', input_shape=(X_treino.shape[1],)),
        Dense(64, activation='relu'),
        Dense(1, activation='sigmoid')
    ])
    modelo.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    
    modelo.fit(X_treino, Y_treino, epochs=15, batch_size=32, validation_data=(X_validacao, Y_validacao), verbose=0)
    
    perda, precisao = modelo.evaluate(X_teste, Y_teste, verbose=0)
    print(f"Treinamento finalizado. Precisão no teste final: {precisao*100:.2f}%")
    
    datasets = {'X_treino': X_treino, 'X_validacao': X_validacao, 'X_teste': X_teste}
    return modelo, datasets

def executar_backtest(modelo, df_modelo, datasets, ticker_principal):
    print("--- Executando backtest com otimização de limite ---")
    if modelo is None or datasets is None: return

    probabilidades_validacao = modelo.predict(datasets['X_validacao'], verbose=0)
    df_validacao = df_modelo.iloc[len(datasets['X_treino']) : len(datasets['X_treino']) + len(datasets['X_validacao'])].copy()
    melhor_retorno = -np.inf
    melhor_limite = 0.5
    capital_inicial = 1000
    for limite in np.arange(0.40, 0.61, 0.01):
        df_validacao['Sinal_IA'] = (probabilidades_validacao > limite).astype(int)
        df_validacao['Retorno_IA'] = df_validacao['Retorno_Futuro'] * df_validacao['Sinal_IA']
        valor_final = capital_inicial * (1 + df_validacao['Retorno_IA']).cumprod().iloc[-1]
        if valor_final > melhor_retorno:
            melhor_retorno = valor_final
            melhor_limite = limite
    print(f"Melhor Limite de Compra encontrado: {melhor_limite:.2f}")

    if len(datasets['X_teste']) == 0:
        print("Aviso: Conjunto de teste vazio.")
        return

    probabilidades_teste = modelo.predict(datasets['X_teste'], verbose=0)
    sinal_predito_teste = (probabilidades_teste > melhor_limite).astype(int).flatten()
    start_index_teste = len(datasets['X_treino']) + len(datasets['X_validacao'])
    df_teste_oficial = df_modelo.iloc[start_index_teste:].copy()
    df_teste_oficial['Sinal_IA_Otimizado'] = sinal_predito_teste
    df_teste_oficial['Retorno_IA'] = df_teste_oficial['Retorno_Futuro'] * df_teste_oficial['Sinal_IA_Otimizado']
    
    retorno_acumulado_ia = capital_inicial * (1 + df_teste_oficial['Retorno_IA']).cumprod()
    valor_final_ia = retorno_acumulado_ia.iloc[-1]
    lucro_ia = valor_final_ia - capital_inicial
    
    retorno_acumulado_benchmark = capital_inicial * (1 + df_teste_oficial['Retorno_Futuro']).cumprod()
    valor_final_benchmark = retorno_acumulado_benchmark.iloc[-1]
    lucro_benchmark = valor_final_benchmark - capital_inicial

    volatilidade_ia = df_teste_oficial['Retorno_IA'].std() * 100
    volatilidade_benchmark = df_teste_oficial['Retorno_Futuro'].std() * 100
    
    plt.figure(figsize=(12, 6))
    plt.plot(retorno_acumulado_ia.index, retorno_acumulado_ia, label='Estratégia IA', color='blue')
    plt.plot(retorno_acumulado_benchmark.index, retorno_acumulado_benchmark, label='Benchmark (Buy & Hold)', color='gray', linestyle='--')
    plt.title(f'Desempenho da Estratégia IA vs. Benchmark para {ticker_principal}')
    plt.xlabel('Data')
    plt.ylabel('Capital Acumulado (R$)')
    plt.legend()
    plt.grid(True)
    formatter = mticker.FormatStrFormatter('R$ %.2f')
    plt.gca().yaxis.set_major_formatter(formatter)
    plt.tight_layout()
    nome_arquivo_grafico = f'desempenho_{ticker_principal}.png'
    plt.savefig(nome_arquivo_grafico)
    print(f"Gráfico de desempenho salvo como: {nome_arquivo_grafico}")
    # plt.show()
    
    print(f"\n--- TESTE DE DESEMPENHO OFICIAL ---")
    print(f"Período Testado: {df_teste_oficial.index.min().strftime('%Y-%m-%d')} a {df_teste_oficial.index.max().strftime('%Y-%m-%d')}")
    print(f"Capital Final (Estratégia IA Otimizada): R$ {valor_final_ia:.2f}")
    print(f"Capital Final (Benchmark 'Buy & Hold'):  R$ {valor_final_benchmark:.2f}")
    print(f"----------------------------------------------------------")
    print(f"Lucro/Perda da Estratégia IA: R$ {lucro_ia:.2f}")
    print(f"Lucro/Perda do Benchmark:      R$ {lucro_benchmark:.2f}")
    print(f"----------------------------------------------------------")
    print(f"Volatilidade (Risco) da Estratégia IA: {volatilidade_ia:.3f}%")
    print(f"Volatilidade (Risco) do Benchmark:   {volatilidade_benchmark:.3f}%")
    print("="*60 + "\n")


# --- ETAPA 4: BLOCO DE EXECUÇÃO PRINCIPAL ---
if __name__ == '__main__':
    # --- Painel de Controle ---
    lista_de_ativos_para_teste = ['VALE3.SA', 'ITUB4.SA', 'LREN3.SA']
    
    DATA_FIM_TREINO = datetime(2023, 12, 31).date()
    DATA_FIM_VALIDACAO = datetime(2024, 9, 30).date()
    
    # --- Laço de Repetição para testar todos os ativos ---
    for ativo_alvo in lista_de_ativos_para_teste:
        
        print("\n" + "="*80)
        print(f" INICIANDO ANÁLISE PARA: {ativo_alvo} ".center(80, "="))
        print("="*80)
        
        config = CONFIGURACAO_ATIVOS.get(ativo_alvo, CONFIGURACAO_ATIVOS["DEFAULT"])
        print(f"Setor: {config['setor']}")

        start_date = DATA_FIM_VALIDACAO - timedelta(days=5*365)
        end_date = datetime.now()
        fatores_base_tickers = ['CL=F', 'BRL=X']

        dados_brutos = coletar_dados(
            ticker_principal=ativo_alvo,
            fatores_base=fatores_base_tickers,
            fatores_especificos=config['fatores_especificos'],
            nomes_fatores=config['nomes_fatores'],
            start=start_date,
            end=end_date
        )
        
        if dados_brutos:
            df_modelo_pronto = preparar_dataframe(dados_brutos)
            
            if df_modelo_pronto is not None:
                colunas_treino = ['Momentum_5d', 'Volatilidade_10d', 'Score_ESG_Politico', 'Preco_Petroleo'] + config['nomes_fatores']
                
                modelo, datasets = treinar_avaliar_modelo(df_modelo_pronto, colunas_treino, DATA_FIM_TREINO, DATA_FIM_VALIDACAO)
                
                executar_backtest(modelo, df_modelo_pronto, datasets, ticker_principal=ativo_alvo)
```
