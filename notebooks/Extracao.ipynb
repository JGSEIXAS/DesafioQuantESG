```python
# Instale as bibliotecas necessárias. 
# Se já estiverem instaladas, esta célula só checa e avança.
!pip install yfinance pandas --user

import yfinance as yf
import pandas as pd
from datetime import datetime, timedelta
```

```python
# 1. Defina o período: 5 anos de dados diários
end_date = datetime.now()
start_date = end_date - timedelta(days=5*365)

# 2. Defina os Tickers (As 10 Ações mais relevantes para um modelo quantitativo/ESG)
tickers = [
    # Commodities/Risco Ambiental Alto
    'VALE3.SA', 'PETR4.SA', 'BEEF3.SA', 
    # Bancos/Financeiro (Grande peso no índice)
    'ITUB4.SA', 'BBDC4.SA', 'BPAC11.SA',
    # Energia/Saneamento (Regulatório/Social)
    'ELET3.SA', 'CMIG3.SA', 
    # Varejo/Outros
    'LREN3.SA', 'WEGE3.SA'
] 

print(f"Iniciando a coleta de dados de 5 anos para {len(tickers)} ações...")

# 3. Baixar os dados
# Isso criará um DataFrame (tabela) com todas as colunas de preço e volume para as 10 ações.
dados_mercado = yf.download(tickers, start=start_date, end=end_date)

# 4. Exibir a estrutura e as primeiras linhas para confirmação
print("--- Dados Históricos Coletados com Sucesso! ---")
print(dados_mercado.head())
print(f"\nNúmero total de dias de trading: {dados_mercado.shape[0]}")
```
